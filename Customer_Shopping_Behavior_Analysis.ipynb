```python
# 1. Import libraries
import pandas as pd
from sqlalchemy import create_engine

# 2. Load dataset
df = pd.read_csv("customer_shopping_behavior.csv")

# 3. Handle missing values
df['Review Rating'] = df['Review Rating'].fillna(
    df.groupby('Category')['Review Rating'].transform('median')
)

# 4. Standardise column names
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

# 5. Rename column
df = df.rename(columns={
    'purchase_amount_(usd)': 'purchase_amount'
})

# 6. Create age groups
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(
    df['age'],
    q=4,
    labels=labels
)

# 7. Map purchase frequency
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df[
    'frequency_of_purchases'
].map(frequency_mapping)

# 8. Drop redundant column
df = df.drop('promo_code_used', axis=1)

# 9. Connect to PostgreSQL
engine = create_engine(
    "postgresql+psycopg2://postgres:SinPost@localhost:5432/customer_behavior"
)

# 10. Upload cleaned data
df.to_sql(
    'customer',
    engine,
    if_exists='replace',
    index=False
)

# 11. Verify
print("Data uploaded successfully!")
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
```

```python
# 1. Import libraries
import pandas as pd
from sqlalchemy import create_engine

# 2. Load dataset
df = pd.read_csv("customer_shopping_behavior.csv")

# 3. Handle missing values
df['Review Rating'] = df['Review Rating'].fillna(
    df.groupby('Category')['Review Rating'].transform('median')
)

# 4. Standardise column names
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

# 5. Rename column
df = df.rename(columns={
    'purchase_amount_(usd)': 'purchase_amount'
})

# 6. Create age groups
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(
    df['age'],
    q=4,
    labels=labels
)

# 7. Map purchase frequency
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df[
    'frequency_of_purchases'
].map(frequency_mapping)

# 8. Drop redundant column
df = df.drop('promo_code_used', axis=1)

# 9. Connect to PostgreSQL
engine = create_engine(
    "postgresql+psycopg2://postgres:SinPost@localhost:5432/customer_behavior"
)

# 10. Upload cleaned data
df.to_sql(
    'customer',
    engine,
    if_exists='replace',
    index=False
)

# 11. Verify
print("Data uploaded successfully!")
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
```
